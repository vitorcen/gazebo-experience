# Worlds 目录 - 世界场景示例

本文档介绍 `examples/worlds/` 目录下的所有 SDF 场景文件，可直接用 `gz sim <文件名>` 运行。

> **提示**：大多数场景都提供了控制命令示例，可通过 `gz topic` 发送指令控制模型。

## 🧹 进程管理
由于使用了后台启动 (`&`)，仿真窗口关闭后后台进程可能仍会驻留。运行下方单元格强制清理。

```python
import os
os.system("pkill -9 -f gz-sim")
print("所有仿真进程已强制清理(SIGKILL)。")
```
